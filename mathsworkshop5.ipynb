```python
import numpy as np
import matplotlib.pyplot as plt
""" Problem 2: Create 2-dimensional position vectors and plot them. """
origin = np.array([0,0])
V = np.array([1,1])
W = np.array([-2.0,-1])
fig, ax = plt.subplots()
# Add vectors V and W to the plot
ax.quiver(origin[0], origin[1], V[0], V[1], angles='xy', scale_units='xy', scale=1, color='r')
ax.quiver(origin[0], origin[1], W[0], W[1], angles='xy', scale_units='xy', scale=1, color='b')
ax.text(origin[0]-0.1, origin[1]+0.1, "O")
ax.text(V[0], V[1], "v")
ax.text(W[0]-0.125, W[1]-0.1, "w")
ax.set_xlim([-3, 3])
ax.set_ylim([-3, 3])
ax.set_xlabel('X')
ax.set_ylabel('Y')
plt.grid()
plt.show()

```

```python
#a
import numpy as np
import matplotlib.pyplot as plt

U = np.array([1, 0.25, 0])
V = np.array([0, 1, 0.25])
W = np.array([0.25, 0, 1])

fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')

origin = [0, 0, 0]

ax.quiver(*origin, *U, color='r')
ax.quiver(*origin, *V, color='b')
ax.quiver(*origin, *W, color='g')

plt.title("3D Vectors")
plt.show()
```

```python
#b
import numpy as np
import matplotlib.pyplot as plt

U = np.array([-0.5, 0.5])
V = np.array([1, 1])

W = U + V

plt.quiver(0, 0, U[0], U[1], angles='xy', scale_units='xy', scale=1, color='r')
plt.quiver(0, 0, V[0], V[1], angles='xy', scale_units='xy', scale=1, color='b')
plt.quiver(0, 0, W[0], W[1], angles='xy', scale_units='xy', scale=1, color='g')

plt.xlim(-1, 2)
plt.ylim(-1, 2)
plt.grid()

plt.title("2D Vector Addition")
plt.show()
```

```python
#c
import numpy as np

def magnitude(vec):
    return np.sqrt(np.sum(vec**2))

U = np.array([1.5, 1.5, 2.0])
print(magnitude(U))
```

```python
#D
def direction_cosines(vec):
    mag = magnitude(vec)
    return vec / mag

O = np.array([1.5, 1.5, 2.0])
print(direction_cosines(O))
```

```python
#e
def dot_product(a, b):
    return np.dot(a, b)

a = np.array([1.5, 1.5, 2.0])
b = np.array([-1.5, 0.5, 2.0])

print(dot_product(a, b))
```

```python
#f
def cross_product(a, b):
    return np.cross(a, b)

a = np.array([1.5, 1.5, 2.0])
b = np.array([-1.5, 0.5, 2.0])

print(cross_product(a, b))
```
