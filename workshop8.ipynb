```python
#example1
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-5, 5, 100)

# System a: x+3y=5, x-2y=-5, x+y=3
y1 = (5 - x) / 3
y2 = (5 + x) / 2
y3 = 3 - x

fig, ax = plt.subplots()
ax.plot(x, y1, color='r')
ax.plot(x, y2, color='g')
ax.plot(x, y3, color='b')

ax.text(0.5, 0.6, '$x+3y = 5$', color='r')
ax.text(-3, 0.5, '$x-2y = -5$', color='g')
ax.text(1.2, 2, '$x+y = 3$', color='b')

ax.set_xlim(-4, 4)
ax.set_ylim(-2, 6)
ax.axvline(color='k', linewidth=1)
ax.axhline(color='k', linewidth=1)
ax.set_xticks(list(range(-4, 5)))
ax.set_aspect('equal')
ax.grid(True, ls=':')
plt.xlabel("X")
plt.ylabel("Y")
plt.show()
# Conclusion: All 3 lines intersect at the same point (-1, 2)
```

Numpy.linalg

linalg is a submodule in NumPy, a popular Python library used for numerical computations, particularly in the domain of linear algebra. linalg stands for "linear algebra". This submodule provides functions and classes for various linear algebra operations, such as matrix decompositions, eigenvalue computations, matrix multiplication, solving linear equations, and many others.

```python
import numpy as np

# Note: worksheet has typos — corrected below
a = np.array([[1, 2], [3, 4]])   # fixed: needs double brackets

# linalg.det — determinant
det_a = np.linalg.det(a)
print("Determinant =", det_a)      # -2.0

# linalg.matrix_rank — rank
rank_a = np.linalg.matrix_rank(a)
print("Rank =", rank_a)             # 2

# linalg.inv — inverse
a_inv = np.linalg.inv(a)
print("Inverse =")
print(a_inv)

# linalg.solve — solve ax = b
# x + 2y = 1
# 3x + 5y = 2
a = np.array([[1, 2], [3, 5]])   # fixed: was [3,4] in worksheet
b = np.array([1, 2])              # fixed: was np.array[1,2]
x = np.linalg.solve(a, b)
print("Solution x =", x)           # [-1.  1.]
```

```python
#Task 3
#1. Create a matrix 𝑨 of order 5 × 5 with random entries. Calculate the rank of the matrix 𝑨 to check the number of linearly independent columns or rows and to check if the matrix is invertible. What are the other ways to check if the matrix is invertible?
import numpy as np

A = np.random.randint(1, 10, (5, 5)).astype(float)
print("Matrix A =")
print(A)
rank = np.linalg.matrix_rank(A)
print(f"Rank = {rank}")
print(f"Is invertible (rank == 5): {rank == 5}")
det = np.linalg.det(A)
print(f"Determinant = {det:.4f}")
print(f"Invertible (det ≠ 0): {abs(det) > 1e-10}")
try:
    A_inv = np.linalg.inv(A)
    print("Inverse exists — matrix IS invertible")
except np.linalg.LinAlgError:
    print("Singular matrix — NOT invertible")
```

```python
#2. Generate a matrix 𝑨 of order 5 × 5 with random entries and generate a vector (column) 𝒙 of length 5 with random entries. Calculate Matrix-vector product 𝒃 = 𝑨𝒙. Now, we have a linear system, i.e. the matrix 𝑨 and a vector 𝒃. Solve this Linear System.
import numpy as np
A = np.random.rand(5, 5)
x = np.random.rand(5)
b = A @ x
print("Matrix A =")
print(A)
print("Vector x (original) =", x)
print("Vector b = Ax =", b)
x_solved = np.linalg.solve(A, b)
print("Solved x =", x_solved)
print("Solution matches original x:", np.allclose(x, x_solved))
```

```python
#3. Solve the following system.

import numpy as np
A = np.array([[ 8,  3, -2],
              [-4,  7,  5],
              [ 3,  4, -12]])

b = np.array([9, 15, 35])
solution = np.linalg.solve(A, b)
print(f"x = {solution[0]:.4f}")
print(f"y = {solution[1]:.4f}")
print(f"z = {solution[2]:.4f}")
print("Verification (A @ solution == b):", np.allclose(A @ solution, b))
```

Task 4:

1. Consider the following 10 points in X-Y plane: (−3, −7.6), (−2,1), (−1, 4.8), (0, 5), (1, 2.8), (2, −0.6), (3, −4), (4, −6.2), (5, −6), (7, 6.4). Plot these points in X-Y plane. Connect the points to guess the polynomial. Once you guess your polynomial, formulate a Linear System and then solve it to find the coefficient of the polynomial. Finally, check if the polynomial fits along with the above given points.

```python
import numpy as np
import matplotlib.pyplot as plt

x_pts = np.array([-3, -2, -1, 0, 1, 2, 3, 4, 5, 7])
y_pts = np.array([-7.6, 1, 4.8, 5, 2.8, -0.6, -4, -6.2, -6, 6.4])


plt.figure(figsize=(10, 5))
plt.subplot(1, 2, 1)
plt.scatter(x_pts, y_pts, color='red', zorder=5)
plt.plot(x_pts, y_pts, 'b--', alpha=0.4)
plt.title("Given Points")
plt.grid(True)
coeffs = np.polyfit(x_pts, y_pts, 3)
print("Polynomial coefficients (highest to lowest degree):", coeffs)

# Step 3: Plot the fitted polynomial
x_line = np.linspace(-4, 8, 200)
y_line = np.polyval(coeffs, x_line)

plt.subplot(1, 2, 2)
plt.scatter(x_pts, y_pts, color='red', zorder=5, label='Data points')
plt.plot(x_line, y_line, 'g-', label='Fitted polynomial')
plt.title("Polynomial Fit")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```
